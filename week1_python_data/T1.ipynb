```python
scores = [45, 82, 67, 90, 38]
```

```python
# List comprehension: filter and transform in one line
passed = [s for s in scores if s >= 50]
curved = [s + 5 for s in scores]
print(passed, curved)
```

```python
#  Dictionary comprehension
students = ["Ali", "Imad", "Hamza"]
lengths = {name: len(name) for name in students}
print(lengths)
```

```python
# Lambda with sorted
people = [("Ali", 25), ("Imad", 22), ("Hamza", 30)]
by_age = sorted(people, key=lambda p: p[1])
print(by_age)
```

```python
# List Comprehension
# Example A: Extracting lowercase domain names from emails
emails = ["User1@Gmail.com", "Admin@Yahoo.COM", "Test@Outlook.com"]
domains = [e.split("@")[1].lower() for e in emails]
print(domains)

# Example B: Filtering out negative numbers and squaring the positive ones

numbers = [4, -2, 7, -9, 3]
squared_positives = [n ** 2 for n in numbers if n > 0]
print(squared_positives)
```

```python
# Dictionary Comprehension
# Example A: Currency / Price Conversion
usd_prices = {"laptop": 1000, "mouse": 25, "keyboard": 75}
rate = 280

local_prices = {item: price * rate for item, price in usd_prices.items()}
print(local_prices)

# Example B: Inverting / Swapping Keys and Values
# You have an ID-to-username lookup dictionary, and you need to reverse it so you can look up an ID by username.
user_roles = {101: "admin", 102: "editor", 103: "viewer"}
role_to_id = {role: uid for uid, role in user_roles.items()}
print(role_to_id)
```

```python
# Practice 1: Numbers & Threshold Filter (Function)
def summarize_scores(scores, passing_grade):
    return {
        "total_students": len(scores),
        "highest_score": max(scores),
        "lowest_score": min(scores),
        "passed_count": len([p for p in scores if p >= passing_grade])
    }

summarize_scores([10, 30, 50 ,100], 50)
```
